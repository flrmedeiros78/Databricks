Build Silver Layer

```python
%sql
USE CATALOG dbacademy;
USE SCHEMA get_started_de;

CREATE OR REPLACE TABLE  employes_silver AS
SELECT 
 ID
,FirstName
,Country
,UPPER(Role) as Role
,current_timestamp() as dt_procs_timestamp
,current_date() AS dt_procs_date
FROM employes_bronze;
```

```python
%sql
SELECT * FROM employes_silver;
```

```python
%sql
CREATE OR REPLACE TEMP VIEW temp_total_employes as 
SELECT
Role,
COUNT(1) AS TotalEmployes
FROM employes_silver
GROUP BY Role;
```

```python
%sql
SELECT * FROM temp_total_employes;
```
