Criando uma tabela Delta utilizando o arquivo employees.csv

```sql
DROP TABLE IF EXISTS employers;
```

```sql
USE CATALOG dbacademy;
USE SCHEMA get_started_de;
```

```sql
SHOW TABLES;
SHOW VOLUMES;
```

```sql
LIST '/Volumes/dbacademy/get_started_de/myfiles'
```

```sql
SELECT * from read_files ('/Volumes/dbacademy/get_started_de/myfiles/employees.csv');
```

```sql
%sql
-- cria tabela employers usando o arquivo employees.csv
CREATE TABLE IF NOT EXISTS delta_employes_01
AS SELECT * from read_files ('/Volumes/dbacademy/get_started_de/myfiles/employees.csv');
```

```sql
%sql
--Consulta Tadela Delta
SELECT * FROM delta_employes_01;
```
