```sql
%sql
USE CATALOG dbacademy;
USE SCHEMA get_started_de;

CREATE OR REFRESH MATERIALIZED VIEW employes_silver_sdp
(
    CONSTRAINT valida_id EXPECT (ID IS NOT null),
    CONSTRAINT valida_name EXPECT(FirstName IS NOT null)
)
COMMENT 'validação para os campos ID, e FirsName ambos Not Null'
AS SELECT 
 ID,
 UPPER(FirstName) AS FirstName,
 UPPER(Country) AS Country, 
 UPPER(Role) AS Role,
 current_timestamp() as dt_procs_timestamp,
 current_date() as dt_procs
FROM employes_bronze_sdp;




```
