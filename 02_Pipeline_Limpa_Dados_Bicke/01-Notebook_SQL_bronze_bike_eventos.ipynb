Ingest data
Create a bronze table that reads directly from the raw CSV source.
In bronze_bike_events.sql, create the bronze streaming table:

```sql
USE CATALOG dbacademy;
USE SCHEMA default;

CREATE OR REFRESH STREAMING TABLE tb_bronze_bike_eventos
COMMENT 'Criando tabela para popular os dados raw da base de dados bike-share-rides.csv'
AS SELECT * 
FROM STREAM(read_files('/Volumes/dbacademy/default/raw_data/rides/build_data_pipeline_demo/',
format =>'csv',
header => 'true',
inferSchema => 'true',
sep => ','));
```
